```julia
using WAVI
using MAT, FileIO  # For loading .mat files
using Printf, Plots
```

```julia
#
#Grid and boundary conditions
# 
nx = 580#162 # use the clipped grid
ny = 719#177 # use the clipped grid
nσ = 26 # tried with less than 26, but could not reach sigma=1.0 and got error. seems to be given by contructor but cannot identify where exactly
x0 = 238000.0#-2260000.0
y0 = -2.5586e6#-1292000.0
dx = 1050.0
dy = 1050.0

# Paths: static fields from original setup, evolved fields from inversion output
static_basepath = "/home/tgia02/repos/WAVIConstructor.jl/examples/outputs/Greenland_CE_1km/"
rosie_inv_inversion_output_path = "/home/tgia02/repos/WAVI.jl/example_drivers/Rosie_examples/Greenland_CE_1km_βpower0.1_Bpower_shelf0.1_Bpower_grounded0.05/"
mismip7_inversion_output_path = "/home/tgia02/repos/WAVI.jl/example_drivers/Rosie_examples/Greenland_CE_1km_βpower0.1_Bpower_shelf0.1_Bpower_grounded0.05_MISMIP7_prep/" 
inversion_iteration = 25
```

```julia
mat_filename = @sprintf("outfile%010d.mat", inversion_iteration)
rosie_inv_mat_filepath = joinpath(rosie_inv_inversion_output_path, mat_filename)
mismip7_mat_filepath = joinpath(mismip7_inversion_output_path, mat_filename)
```

```julia
rosie_inv_mat_data = matread(rosie_inv_mat_filepath)
mismip7_mat_data = matread(mismip7_mat_filepath)
```

```julia
JRMS1 = rosie_inv_mat_data["JRMS"]
JRMS2 = mismip7_mat_data["JRMS"]

plot(JRMS1)
plot!(JRMS2)
```

```julia
JKV1 = rosie_inv_mat_data["JKV"]
JKV2 = mismip7_mat_data["JKV"]

plot(log10.(JKV1))
plot!(log10.(JKV2))
```

```julia
field = "beta"

f1 = rosie_inv_mat_data[field]
f2 = mismip7_mat_data[field]
diff = f1 .- f2

println("rosie_inv: $(extrema(f1))")
println("mismip7:   $(extrema(f2))")
println("diff:      $(extrema(diff))")

p1 = heatmap(log10.(f1)',
    title = "rosie_inv",)

p2 = heatmap(log10.(f2)',
    title = "mismip7",)

# p3 = heatmap(log10.(diff),
#     title = "diff")

plot(p1, p2, layout = (1, 2), size = (900, 400))
```
