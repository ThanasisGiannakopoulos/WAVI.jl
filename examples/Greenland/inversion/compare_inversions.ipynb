```julia
using WAVI
using MAT, FileIO  # For loading .mat files
using Printf, Plots
using NCDatasets
```

```julia
path = "/home/tgia02/repos/WAVIConstructor.jl/examples/outputs/Greenland_Sorgenfri_2022_150.0m_basins_150m_res_min_thick10.0"

input = NCDataset(path*"/wavi_input.nc")

#Grid and boundary conditions
nx = length(input["x"])
ny = length(input["y"])
nσ = 26
dx = abs(input["x"][2] - input["x"][1])
dy = abs(input["y"][2] - input["y"][1])
x0 = input["x"][1] - dx/2
y0 = input["y"][1] - dy/2


static_basepath = path

inversion_output_path1 = "/home/tgia02/repos/WAVI.jl/example_drivers/Rosie_examples/Greenland_Sorgenfri_2022_150.0m_basins_150m_res_min_thick10.0_βpower0.1_Bpower_shelf0.1_Bpower_grounded0.05_ISMIP7_prep_run2/"
inversion_output_path2 = "/home/tgia02/repos/WAVI.jl/example_drivers/Rosie_examples/Greenland_Sorgenfri_2022_150.0m_basins_150m_res_min_thick10.0_βpower0.1_Bpower_shelf0.1_Bpower_grounded0.05_ISMIP7_prep_decoupled_damage_g3d_preBfactor_vertical_uniform/"

inversion_iteration = 25  # UPDATE THIS: which iteration to load (last iteration = 25 for your setup)
```

```julia
udata = coalesce.(input["udata_mask"][:,:].*input["udata"][:,:], 0.0)
vdata = coalesce.(input["vdata_mask"][:,:].*input["vdata"][:,:], 0.0)
```

```julia
mat_filename = @sprintf("outfile%010d.mat", inversion_iteration)
mat_filepath1 = joinpath(inversion_output_path1, mat_filename)
mat_filepath2 = joinpath(inversion_output_path2, mat_filename)
```

```julia
mat_data1 = matread(mat_filepath1)
mat_data2 = matread(mat_filepath2)
```

```julia
JRMS1 = mat_data1["JRMS"]
JRMS2 = mat_data2["JRMS"]
plot(JRMS1)
plot!(JRMS2)
```

```julia
field = "u"

f1 = udata#mat_data1[field]  # Thickness (nx, ny)
println("f1 extrema = ", extrema(f1))
println("f1 sum = ", sum(abs.(f1)))

f2 = mat_data2[field]  # Thickness (nx, ny)
println("f2 extrema = ", extrema(f2))
println("f2 sum = ", sum(abs.(f2)))

abs_diff = abs.(f2 .- f1)
println("abs diff extrema = ", extrema(abs_diff))
println("abs diff sum = ", sum(abs_diff))

p1 = heatmap(
    log10.(abs.(f1))',
    title = "log10(f1)",
    colorbar = true
)

p2 = heatmap(
    log10.(abs.(f2))',
    title = "log10(f2)",
    colorbar = true
)

p3 = heatmap(
    log.(abs.(f2 .- f1))',
    title = "log10(abs(f2-f1))",
    colorbar = true
)

plot(p1, p2, p3, layout = (1, 3), size = (1200, 350))
```
