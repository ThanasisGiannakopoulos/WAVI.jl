```julia
using WAVI
using MAT, FileIO
using Printf, Plots, NCDatasets
```

```julia
path = "/home/tgia02/repos/WAVIConstructor.jl/examples/outputs/Greenland_Sorgenfri_2022_150.0m_basins_150m_res_min_thick10.0"

input = NCDataset(path*"/wavi_input.nc")

#Grid and boundary conditions
nx = length(input["x"])
ny = length(input["y"])
nσ = 26
dx = abs(input["x"][2] - input["x"][1])
dy = abs(input["y"][2] - input["y"][1])
x0 = input["x"][1] - dx/2
y0 = input["y"][1] - dy/2

# Paths: static fields from original setup, evolved fields from inversion output
output_path = "./output/Sorgenfri_2022_150.0m_basins_150m_res_min_thick10.0_decoupled_damage_g3d_preBfactor_vertical_uniform_relaxation_out_dt_1day_instant_healing/"
iteration = 30
```
